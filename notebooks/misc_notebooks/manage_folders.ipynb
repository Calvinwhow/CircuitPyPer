```python
from pathlib import Path
import gzip
import shutil

base_directory = Path("/Volumes/HowExp/datasets/01a_ADNI_Alzheimer_Atrophy/neuroimaging/all_patients")

created = 0
skipped = 0

for gz_path in base_directory.glob("sub-*/ses-01/mri/y_*.nii.gz"):
    nii_path = gz_path.with_suffix("")  # removes .gz -> .nii
    if nii_path.exists():
        skipped += 1
        continue

    nii_path.parent.mkdir(parents=True, exist_ok=True)
    with gzip.open(gz_path, "rb") as f_in, open(nii_path, "wb") as f_out:
        shutil.copyfileobj(f_in, f_out)

    created += 1

print(f"created={created} skipped={skipped}")
```

```python
import shutil
from pathlib import Path

origin = Path("/Volumes/HowExp/datasets/03y_Cologne_AD_Fornix_DBS/data2share")
dest_dir = Path("/Volumes/HowExp/datasets/03y_Cologne_AD_Fornix_DBS/BIDS")

for ses_dir in origin.glob("sub-*/ses-*"):
    if not ses_dir.is_dir():
        continue

    sub = ses_dir.parent.name      # e.g. sub-pat001
    ses = ses_dir.name             # e.g. ses-01

    dest_ses_dir = dest_dir / sub / ses
    dest_ses_dir.mkdir(parents=True, exist_ok=True)

    for item in ses_dir.iterdir():
        dest_item = dest_ses_dir / item.name

        if item.is_dir():
            shutil.copytree(item, dest_item, dirs_exist_ok=True)
            print(f"Copied folder: {item} -> {dest_item}")

        elif item.is_file():
            shutil.copy2(item, dest_item)
            print(f"Copied file: {item} -> {dest_item}")
```
