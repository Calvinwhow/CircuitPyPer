```python
from pathlib import Path
import shutil
import gzip
dest_root = Path('/Volumes/OneTouch/datasets/01z_SungshinKim_ParietalTMS-Atrophy/BIDS')
orig_root = Path('/Volumes/HowExp/datasets/06a_Jung_Alzheimer_TMS/BIDS')


for f in orig_root.rglob("*MNI152_T1_2mm*"):
    if f.name.startswith("._") or f.name.startswith("."):
        continue
    
    subject = f.name.split('sub-')[1].split('-MNI')[0]
    subfolder = f.parent.name
    session = 'pre'
    
    dest_folder = dest_root / f"sub-{subject}" / f"ses-{session}" / subfolder
    dest_folder.mkdir(parents=True, exist_ok=True)
    
    dest_file = dest_folder / f.name  # keeps .nii.gz
    shutil.copy2(f, dest_file)
    
```
