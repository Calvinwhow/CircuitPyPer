````python
import numpy as np

x = df[[
    "Total Turning Time (% Baseline)",
    "RetestTotal Turning Time (% Baseline)"
]].dropna().to_numpy(dtype=float)

n, k = x.shape
grand = x.mean()
row_means = x.mean(axis=1, keepdims=True)
col_means = x.mean(axis=0, keepdims=True)

MSR = k * np.sum((row_means - grand)**2) / (n - 1)
MSC = n * np.sum((col_means - grand)**2) / (k - 1)
MSE = np.sum((x - row_means - col_means + grand)**2) / ((n - 1) * (k - 1))

icc = (MSR - MSE) / (MSR + (k - 1)*MSE + k*(MSC - MSE)/n)
print(f"ICC(2,1) = {icc:.4f}")
from scipy.stats import f

F = MSR / MSE
df1 = n - 1
df2 = (n - 1) * (k - 1)

p = f.sf(F, df1, df2)

print(f"ICC(2,1) = {icc:.4f}, F({df1}, {df2}) = {F:.3f}, p = {p:.4g}")
````
