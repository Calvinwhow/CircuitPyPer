```python
import pandas as pd
from calvin_utils.plotting_utils.simple_dynamite_plot import SimpleDynamitePlotWrapper
# Load your CSV (wide format)
csv_path = '/Users/cu135/Partners HealthCare Dropbox/Calvin Howard/studies/acoe_reliability/scores_v1.csv'
df = pd.read_csv(csv_path)

pairs = [
    ('Q3A', 'Q3M'), ('Q4A', 'Q4M'), ('Q5A', 'Q5M'), ('Q6A', 'Q6M'),
    ('Q7A', 'Q7M'), ('Q8A', 'Q8M'), ('Q9A', 'Q9M'), ('Q10A', 'Q10M'),
    ('Q11A', 'Q11M'), ('Q12A', 'Q12M'), ('Q13A', 'Q13M'), ('Q14A', 'Q14M'),
    ('Q15A', 'Q15M'), ('Q16A', 'Q16M'), ('Q17A', 'Q17M'), ('Q18A', 'Q18M'),
    ('Q19A', 'Q19M'), ('Q20A', 'Q20M'), ('TotalA', 'TotalM'),
]

for qa_col, qm_col in pairs:
    qm_max = df[qm_col].max()
    if qm_max > 0:
        df[qa_col] = df[qa_col] / qm_max
        df[qm_col] = df[qm_col] / qm_max
    else:
        df[qa_col] = 0.0
        df[qm_col] = 0.0

plotter = SimpleDynamitePlotWrapper(df)

plotter.plot(
    columns=[('Q3A', 'Q3M'), ('Q4A', 'Q4M'), ('Q5A', 'Q5M'), ('Q6A', 'Q6M'), ('Q7A', 'Q7M'), ('Q8A', 'Q8M'), ('Q9A', 'Q9M'), ('Q10A', 'Q10M'), ('Q11A', 'Q11M'), ('Q12A', 'Q12M'), ('Q13A', 'Q13M'), ('Q14A', 'Q14M'), ('Q15A', 'Q15M'), ('Q16A', 'Q16M'), ('Q17A', 'Q17M'), ('Q18A', 'Q18M'), ('Q19A', 'Q19M'), ('Q20A', 'Q20M'), ('TotalA', 'TotalM')],
    group_labels=['Q3', 'Q4', 'Q5', 'Q6', 'Q7', 'Q8', 'Q9', 'Q10', 'Q11', 'Q12', 'Q13', 'Q14', 'Q15', 'Q16', 'Q17', 'Q18', 'Q19', 'Q20', 'Total'],
    pair_names=['Auto', 'Manual'],
    dataset_name='Pairwise Dynamite Plot',
    xlabel='Metric',
    ylabel='Value'
)
```
