## ParcelwiseYabPlot

`ParcelwiseYabPlot` plots already-computed parcelwise maps with `yabplot`.

It does **not** calculate parcel scores. The usual workflow is:

1. Build/save a parcel-filled NIfTI with `ParcelwiseDamageMap`.
2. Pass that saved NIfTI to `ParcelwiseYabPlot`.
3. Optionally pass the parcel atlas glob to render one mesh per parcel with outlines.

### Recommended: outlined parcel mesh plot

This mode uses the saved parcelwise NIfTI for values and the parcel mask files for geometry. It is usually better than voxelwise isosurface plotting for parcellated maps.

```python
from calvin_utils.plotting_utils import ParcelwiseYabPlot

plotter = ParcelwiseYabPlot(
    map_path="/path/to/parcelwise_map.nii.gz",
    parcel_path="/path/to/parcels/*.nii.gz",
    out_dir="/path/to/output_dir",
    output_name="my_parcelwise_map",
)

plotter.run(
    views=["superior"],
    cmap="coolwarm",
    edge_mode="feature",
    line_width=0.6,
    smooth_i=8,
    bmesh_alpha=0.03,
)
