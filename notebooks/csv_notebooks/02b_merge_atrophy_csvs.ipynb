```python
from glob import glob
import pandas as pd

root = '/Volumes/OneTouch/datasets/01z_SungshinKim_ParietalTMS-Atrophy/BIDS/sub-*/ses-pre/measurements/regional_atrophy_coarse.csv'

df = pd.DataFrame()

for file in glob(root):
    sub = file.split("sub-")[1].split("/")[0]
    file_df = pd.read_csv(file)

    cols = file_df["ROI"].tolist()

    if df.empty:
        df = pd.DataFrame(columns=["subject"] + cols)

    row_idx = len(df)
    df.loc[row_idx, "subject"] = sub

    for col in cols:
        val = file_df.loc[file_df["ROI"] == col, "Atrophy_Z"].values[0]
        df.loc[row_idx, col] = val

display(df)
df.to_csv("/Volumes/OneTouch/datasets/01z_SungshinKim_ParietalTMS-Atrophy/metadata/atrophy.csv")
```
